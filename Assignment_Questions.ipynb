You are building a sentiment analysis system for customer reviews from an e-commerce platform. The reviews contain:

Design a comprehensive text pre-processing pipeline. For each step (tokenization, normalization, stop-word removal, stemming/lemmatization), explain:

## Answer 2: Pre-processing Pipeline Design

### 1. Normalization (Cleaning and Standardizing)
  * **Why Include this?**
    The data has slang, typos, and product codes. If normalization is not done, the model will think "gonna," "going to," and "Gonna" are three different words. Normalization groups similar concepts, which is essential.
  * **Specific Considerations:**
      * **Lowercase:** Convert everything to lowercase (`"Love"` -\> `"love"`). This is a simple, standard step.
      * **Slang & Contractions:** Use a dictionary to expand slang (`"u"` -\> `"you"`) and contractions (`"didn't"` -\> `"did not"`). This is *vital* for catching negations.
      * **Emojis:** Emojis are pure sentiment\! Not removing them. Instead, I'll convert them to their text meaning (e.g., "😊" -\> " positive\_emoji " or " smiley\_face "). This turns a special character into a useful token.
      * **Noise Removal:** Product codes (`"XF-100"`) and prices (`"₹15000"`) are noise for sentiment. Remove them using regular expressions (regex).
      * **Misspellings:** Try to include a spell-correction step (e.g., correcting `"terible"` to `"terrible"`). This helps the model recognize the word's negative sentiment.
      * **Punctuation:** Remove most punctuation, but *after* handling emojis and contractions.

### 2. Tokenization
  * **Why Include this?**
    Tokenization breaks a sentence into a list of individual words (tokens). We can't do anything else (like remove stop words or lemmatize) until we have these word-level tokens.
  * **Specific Considerations:**
    A simple `text.split(" ")` is not good enough. A proper tokenizer from **spaCy** or **NLTK** is better because it knows how to handle things like "did\_not" (from our contraction step) or hyphenated words.

### 3. Stop-Word Removal
  * **Why Include this?**
    Removing words like "the," "is," and "at" can reduce noise and focus the model on important words ("amazing," "awful").
  * **Specific Considerations:**
    This is the trickiest part for sentiment. The default stop-word list in NLTK includes **"not"**. If we remove it, "This is **not** good" becomes "good," which is the *exact opposite* sentiment.
      * **My Solution:** I will get the standard NLTK stop-word list and then **manually remove all negation words** from it ("not," "no," "n't," "don't," etc.). I'll use this *customized* stop-word list.

### 4. Stemming vs. Lemmatization
  * **Why Include Lemmatization and Exclude Stemming?**
      * **Stemming** (e.g., PorterStemmer) is a crude process that just chops off word endings. "terrible" might become "terribl." This is fast, but "terribl" isn't a real word, and it might lose its clear negative meaning.
      * **Lemmatization** is a smarter, dictionary-based process. It finds the root word ("lemma"). "loved" becomes "love," "worst" becomes "bad," and "better" becomes "good."
  * **Specific Considerations:**
    For sentiment, meaning is key. Lemmatization preserves the dictionary word, which is much better for a model to learn sentiment from. It's slower than stemming, but I think the accuracy gain is worth it.

### The Order of Operations (and Why It Matters)
The order is critical because the steps depend on each other. My final pipeline would be:
1.  **Normalization (Part 1):** On the raw text string.
      * `Lowercase`
      * `Expand Contractions`
      * `Convert Emojis to Text`
      * `Expand Slang`
      * `Remove Noise` (prices, product codes)
      * `Remove Punctuation`
      * `Spell Correction`
2.  **Tokenization:** Split the clean string from Step 1 into a list of tokens.
3.  **Stop-Word Removal:** Loop through the tokens and remove any that are in my *custom* (negation-safe) stop-word list.
4.  **Lemmatization:** Loop through the remaining tokens and find the lemma for each one.

**Why this order?**
  * You **must** clean the text (like expanding "gonna" to "going to") *before* you lemmatize, otherwise, the lemmatizer won't know what "gonna" is.
  * You **must** tokenize *before* you can remove stop words, as stop words *are* tokens.
  * It's much more efficient to lemmatize *after* removing stop words, so you're only processing the words that matter.

````python
### Code Example (using NLTK)
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

# --- One-time downloads (needed for NLTK) ---
# nltk.download('punkt')
# nltk.download('stopwords')
# nltk.download('wordnet')
````

````python
# --- 1. Setup (do this once) ---

# My custom slang/emoji map
special_handling_map = {
    "😊": " positive_emoji ",
    "😠": " negative_emoji ",
    "gonna": " going to ",
    "wanna": " want to ",
    "u": " you ",
    "terible": " terrible " # Simple spell correction
    # ...add more as needed
}

# My custom stop-word list (keeping negations!)
stop_words = set(stopwords.words('english'))
negations = {'not', 'no', 'n\'t', 'don\'t', 'isn\'t', 'wasn\'t', 'aren\'t', 'weren\'t', 'haven\'t', 'hasn\'t', 'hadn\'t'}
stop_words = stop_words - negations # Remove negations from the set

# Initialize lemmatizer
lemmatizer = WordNetLemmatizer()
````

````python
# --- 2. The Pipeline Function ---

def preprocess_review(text):
    
    # --- Step 1: Normalization (Part 1) ---
    
    # Lowercase
    text = text.lower()
    
    # Expand slang, emojis, common typos
    for key, value in special_handling_map.items():
        text = text.replace(key, value)
        
    # Expand contractions (a simple example)
    text = text.replace("n't", " not")
    
    # Remove noise (prices, product codes with regex)
    # This removes prices like ₹1500.50 or 1500.50
    text = re.sub(r'\₹\d+(\.\d{1,2})?', ' ', text) 
    # This removes product codes like XF-100 or 45A-B
    text = re.sub(r'[A-Z0-9]+-[A-Z0-9]+', ' ', text)
    
    # Remove punctuation
    text = re.sub(r'[^\w\s]', ' ', text)
    
    # --- Step 2: Tokenization ---
    # We tokenize *after* cleaning so "going to" becomes two tokens
    tokens = word_tokenize(text)
    
    # --- Steps 3 & 4: Stop-Word Removal and Lemmatization ---
    clean_tokens = []
    for token in tokens:
        # Check if it's NOT a stop word
        if token not in stop_words:
            # If not, lemmatize it and add to our list
            lemma = lemmatizer.lemmatize(token)
            clean_tokens.append(lemma)
            
    # Join tokens back into a clean string
    return " ".join(clean_tokens)
````

````python
# --- 3. Example Usage ---
review = "This XF-100 is terible!! 😠 I'm not gonna buy from u again. Wasted ₹1500.50"

processed_review = preprocess_review(review)

print(f"Original: {review}")
print(f"Processed: {processed_review}")
````

Consider these sentences:

Apply both stemming (Porter Stemmer) and lemmatization to these sentences. Then:

## Answer 3: Stemming vs Lemmatization Trade-offs

````python
import nltk
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
from nltk.corpus import wordnet

# --- One-time downloads ---
# nltk.download('punkt')
# nltk.download('wordnet')
# nltk.download('averaged_perceptron_tagger') # Needed for POS tagging
````

````python
# --- Setup ---
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

sentences = [
    "The meeting was well organized and the organizers did a great job.",
    "She is better at organizing than her predecessor was."
]

# Helper function to get the correct POS tag for the lemmatizer
def get_wordnet_pos(pos_tag):
    if pos_tag.startswith('J'):
        return wordnet.ADJ
    elif pos_tag.startswith('V'):
        return wordnet.VERB
    elif pos_tag.startswith('N'):
        return wordnet.NOUN
    elif pos_tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN # Default to noun
````

````python
print("--- Analysis ---")

for s in sentences:
    print(f"\nORIGINAL: {s}")
    
    tokens = word_tokenize(s)
    
    # --- Stemming ---
    stemmed = [stemmer.stem(t) for t in tokens]
    print(f"STEMMED:  {' '.join(stemmed)}")
    
    # --- Lemmatization (with POS tagging) ---
    pos_tags = nltk.pos_tag(tokens)
    lemmatized = [lemmatizer.lemmatize(t, get_wordnet_pos(pos)) for t, pos in pos_tags]
    print(f"LEMMATIZED: {' '.join(lemmatized)}")
````

### Differences
  * **Stemming (Porter):** This is a crude, rule-based "chopper."
      * It cuts "meeting" to **`meet`**.
      * It cuts "organized," "organizers," and "organizing" all down to **`organ`**. This is a problem, as "organ" isn't the right root and loses the meaning.
      * It cuts "was" to **`wa`**, which isn't a real word.
      * It completely failed on **`better`**, leaving it unchanged.
  * **Lemmatization (WordNet):** This is a "smarter," dictionary-based approach.
      * It knows "was" and "is" are forms of the verb "to be," so both become **`be`**.
      * It correctly identifies "organizers" as a noun and finds the singular root **`organizer`**.
      * It knows "did" is the past tense of "do," so it becomes **`do`**.
      * Most importantly, it knows that "better" is the comparative form of "well" (when tagged as an adjective) and converts it to **`well`**.

**In short:** Use **stemming** when you need a fast and simple way to group words, and you don't care if the result is a real word. Use **lemmatization** when you need the *actual meaning* of the word to be preserved, which is almost always better for high-quality analysis.

### Impact on Systems
  * **Search Engines / Information Retrieval:**
      * **Stemming** is often used here. Why? Speed and **Recall**. If you search for "organizing a meeting," you *also* want documents that say "organized a meeting." Stemming both to `organ` and `meet` ensures you get all related documents, even if the precision isn't perfect. A faster, broader search is the goal.
  * **Text Classification (e.g., Sentiment Analysis):**
      * **Lemmatization** is strongly preferred.
      * **Impact:** The model needs to know the *meaning*. In our example, `better` was changed to `well`. This is a huge, positive signal for a sentiment model. Stemming missed this entirely. If we use stemming, our model has to learn that `better` is positive, `good` is positive, `best` is positive and `well` is positive, all as separate words. With lemmatization, it just has to learn that `well` is positive, which is much more efficient and accurate.

## Answer 4: POS Tagging for Ambiguity Resolution
The sentence is: **"The duck is ready to eat."** 

This sentence shows the **limits of POS tagging**. The ambiguity is:
1.  **Meaning 1:** The duck (an animal) is hungry and wants to eat.
2.  **Meaning 2:** The duck (a meal) is cooked and ready to be eaten.

If we run a POS tagger on this, it will almost certainly produce:
`[('The', 'DT'), ('duck', 'NN'), ('is', 'VBZ'), ('ready', 'JJ'), ('to', 'TO'), ('eat', 'VB'), ('.', '.')]`

Here's the problem: **this tagging is correct for both meanings.**
* In both cases, "duck" is a Noun (NN).
* In both cases, "eat" is a Verb (VB).

The ambiguity here isn't about whether "duck" is a noun or a verb (like in "They can fish"). The ambiguity is **semantic**—it's about the *role* of the noun "duck."
* Is "duck" the **agent** of the verb "eat" (the one doing the eating)?
* Or is "duck" the **patient** of the verb "eat" (the one being eaten)?

Simple POS tagging is not powerful enough to figure this out. It can't resolve this particular ambiguity because the grammatical structure is the same for both interpretations. This requires a more advanced NLP task like **Semantic Role Labeling (SRL)** or using a large language model with real-world knowledge.

````python
import nltk
# nltk.download('punkt')
# nltk.download('averaged_perceptron_tagger')

# Sentence 2: "They can fish."
# Ambiguity: Are they *able* to fish (verb), or do they *put fish into* cans (noun)?

# Meaning 1: They are able to fish.
# Meaning 2: They put fish into cans.
tokens1 = ['They', 'can', 'fish']
# We get the tags: PRP (Pronoun), MD (Modal Verb), VB (Verb)
print(f"'They can fish.' -> {nltk.pos_tag(tokens1)}")
# A tagger *could* produce: PRP (Pronoun), VB (Verb), NN (Noun)
# But this is *much less likely* in English.
# The NLTK tagger is probabilistic and confidently chooses Meaning 1.

# Sentence 3: "Time flies like an arrow."
# Ambiguity: Does time move (verb), or do "time flies" (noun) exist?

# Meaning 1: Time moves quickly.
tokens3 = ['Time', 'flies', 'like', 'an', 'arrow']
# We get the tags: NN (Noun), VBZ (Verb), IN (Preposition), ...
print(f"'Time flies...' -> {nltk.pos_tag(tokens3)}")
````

As you can see, the tagger resolved the ambiguity:
  * In "They can fish," it tagged `can` as a Modal Verb (MD), which forces `fish` to be the main Verb (VB). This locks in the "able to fish" meaning.
  * In "Time flies...," it tagged `flies` as a Verb (VBZ), locking in the "time moves" meaning.

### Rule-Based vs. Probabilistic Approaches
  * **Rule-Based Taggers:**
      * These use a large, hand-written list of rules created by linguists.
      * **Example Rule:** "If a word ends in '-ed', tag it as a Past Tense Verb (VBD), *unless* it is 'bed'."
      * It's like a strict grammar teacher who has a rule for everything.
  * **Probabilistic Taggers (like a Hidden Markov Model - HMM):**
      * These learn from a large, pre-tagged dataset (a corpus). They don't use explicit rules.
      * They calculate the probability of a tag sequence based on context. They ask two questions:
        1.  **Emission:** How likely is the word "fish" to be a `NN` vs. a `VB` in the training data?
        2.  **Transition:** If the *last* word was tagged `MD` (like "can"), what's the probability the *next* word is tagged `VB` (like "fish")?
      * It's like a data-driven detective that picks the *most likely* suspect (tag) based on all the evidence.


### Which Approach Performs Better?
**A probabilistic approach would perform better for all three sentences.**
  * **Why?** All three sentences are ambiguous *because* the context is so important.
  * **"The duck is ready to eat."** A probabilistic tagger knows that the sequence `...to/TO eat/VB` (Verb) is thousands of times more common than `...to/TO eat/NN` (Noun).
  * **"They can fish."** A rule-based tagger might get confused by "can" (it can be a noun or a verb). A probabilistic HMM tagger is *extremely* confident that the sequence `Pronoun -> "can"` almost always means "can" is a Modal Verb (MD), not a noun or a main verb.
  * **"Time flies like an arrow."** This is the classic example. A simple rule-based tagger might see "flies" and tag it as a plural Noun (NNS). A probabilistic tagger sees the *whole sentence* and calculates that the probability of `Time/NN flies/VBZ ...` is vastly higher than `Time/NN flies/NNS ...`.


### Limitations of Both Approaches
  * **Rule-Based Taggers:**
      * **Brittle:** They break as soon as they see a new word, slang, or a typo.
      * **Costly:** It takes an incredible amount of time and linguistic expertise to write and maintain all the rules.
      * **Rule Conflicts:** Rules can contradict each other, leading to errors.
  * **Probabilistic Taggers:**
      * **Data-Hungry:** They are useless without a massive, high-quality, manually tagged corpus to learn from.
      * **Black Box:** It can be hard to know *why* it made a specific tagging decision.
      * **Context Window:** A simple HMM only looks at the previous tag. It can get confused by very long-distance dependencies in a sentence. (Newer models like LSTMs/Transformers are much better at this).

You need to build an NER system for extracting information from medical reports. The text contains:

Compare dictionary-based and CRF-based NER methods for this application:

## Answer 5: Named Entity Recognition System Design

### Advantages and Disadvantages
| Method | Advantages | Disadvantages |
| :--- | :--- | :--- |
| **Dictionary-Based** | 🟢 Simple, fast, and very accurate for known terms.<br>🟢 Great for standardized lists (e.g., all known diseases). | 🔴 **Fails completely** on any new or unseen term (like a new drug name).<br>🔴 Can't handle typos or variations.<br>🔴 Dumb to context (might tag "Dr. Brown" as a color). |
| **CRF-Based**<br>(Conditional Random Fields) | 🟢 **Can find new entities** it hasn't seen before by learning patterns.<br>🟢 Uses context (e.g., "prescribed [DRUG]").<br>🟢 Very robust and powerful. | 🔴 Needs a lot of expensive, manually-labeled training data.<br>🔴 Much more complex and slower to build and train. |


### How to Handle New Drug Names
  * **Dictionary:** It can't. You'd have to **manually update your dictionary** with the new drug name, which is slow and impossible to keep up with.
  * **CRF:** This is its main strength. It learns **patterns** from the data. It will see the new drug name and recognize it based on features like:
      * "It's capitalized."
      * "It appears after the word 'prescribed'."
      * "It's followed by a dosage like '50mg'."
        It can then **infer** that this new, unknown word is a `MEDICATION`.


### Features for a CRF Model
To help the CRF learn these patterns, we don't just give it the word. We give it features *about* the word and its neighbors:
  * **Word Shape:** Is the word capitalized (`Xxxxx`)? All caps (`XXXX`)? Does it have numbers and letters (`XXXX-dd` like "COVID-19")?
  * **Suffix/Prefix:** Does the word end in "-in" (like "Metformin") or "-pam" (like "Diazepam")?
  * **Context:** What are the words immediately before and after it? (e.g., `previous_word="prescribed"`).
  * **POS Tag:** Is this word a Noun?
  * **Dictionary Check:** Is this word in a known list of drugs? (This is a key feature).


### How to Combine Both (Optimal Results)
You use a **hybrid approach:** The dictionary becomes a **feature for the CRF model**. This is called a **"gazetteer" feature**.
For every word in your text, you add a feature that's either `True` or `False`: "Is this word found in my master dictionary of drugs?"

**Why this is the best solution:**
  * For common drugs ("Metformin"), the CRF gets a very strong `True` signal from this feature and will tag it correctly (high accuracy).
  * For new, unseen drugs, this feature will be `False`, but the CRF can **still** identify it as a drug using all its other features (like word shape and context).

This gives you the best of both worlds: the accuracy of a dictionary and the flexibility of a machine learning model.

a) Build a bigram language model and calculate probabilities for:

## Answer 6: N-gram Language Models and Perplexity

````python
import nltk
from nltk.lm.preprocessing import padded_everygram_pipeline
from nltk.lm import MLE # This is the basic, unsmoothed model
from nltk.tokenize import word_tokenize
from nltk.util import ngrams
````

````python
# --- 1. Setup the Corpus ---
corpus_text = [
    "I love machine learning",
    "I love deep learning",
    "Machine learning is fascinating",
    "Deep learning is powerful"
]
# Tokenize the corpus (and lowercase)
tokenized_corpus = [word_tokenize(s.lower()) for s in corpus_text]
````

````python
# --- 2. Build the Bigram Model ---
n = 2 # n=2 for bigrams
# This pads sentences with <s> and </s> and creates the vocabulary
train_data, padded_vocab = padded_everygram_pipeline(n, tokenized_corpus)

# We use the Maximum Likelihood Estimator (MLE)
# This model just calculates P(b|a) = count(a,b) / count(a)
model = MLE(n)
model.fit(train_data, padded_vocab)
````

````python
# --- 3a. Calculate Probability: "I love natural learning" ---
print("--- 'I love natural learning' ---")
test_text_1 = "I love natural learning"
test_tokens_1 = word_tokenize(test_text_1.lower())

# We must get the bigrams for the test sentence
# (<s>, i), (i, love), (love, natural), (natural, learning), (learning, </s>)
test_bigrams_1 = list(ngrams(test_tokens_1, n, 
                             pad_left=True, pad_right=True, 
                             left_pad_symbol="<s>", right_pad_symbol="</s>"))

total_prob_1 = 1
for bigram in test_bigrams_1:
    history = bigram[0] # The "given" word (w_i-1)
    word = bigram[1]    # The word to predict (w_i)
    
    # model.score() gets the probability
    prob = model.score(word, [history])
    print(f"P({word} | {history}) = {prob}")
    total_prob_1 *= prob

print(f"\nTotal Probability = {total_prob_1}")
print("This is 0 because P(natural | love) was 0.")
````

````python
# --- 3b. Calculate Probability: "Machine learning is powerful" ---
print("\n--- 'Machine learning is powerful' ---")
test_text_2 = "Machine learning is powerful"
test_tokens_2 = word_tokenize(test_text_2.lower())

# (<s>, machine), (machine, learning), (learning, is), (is, powerful), (powerful, </s>)
test_bigrams_2 = list(ngrams(test_tokens_2, n, 
                             pad_left=True, pad_right=True, 
                             left_pad_symbol="<s>", right_pad_symbol="</s>"))

total_prob_2 = 1
for bigram in test_bigrams_2:
    history = bigram[0]
    word = bigram[1]
    prob = model.score(word, [history])
    print(f"P({word} | {history}) = {prob}")
    total_prob_2 *= prob

print(f"\nTotal Probability = {total_prob_2}")
print(f"(Note: 1/16 = {1/16})")
````

### The Zero-Probability Problem and Perplexity
  * **The Zero-Probability Problem:**
    This is what we just saw. The bigram `(love, natural)` never appeared in our training data. Therefore, the model assigns it a probability of 0. Since all probabilities are multiplied together, the entire sentence gets a probability of 0, even if it's a perfectly good sentence. This breaks the model.
  * **How Laplace Smoothing Addresses It:**
    Laplace (or "add-one") smoothing fixes this by **adding 1 to every n-gram count**. It's like "pre-seeding" the model with one of every possible bigram. This borrows a tiny bit of probability from n-grams we *have* seen and gives it to the ones we *haven't*.
      * The probability for `(love, natural)` would go from $0/2$ to $(0+1)/(2+V)$ (where V is the vocabulary size). This is a small, non-zero number that prevents the whole chain from becoming 0.
  * **The Concept of Backoff:**
    Backoff is another, slightly smarter, way to handle this. The idea is:
    1.  Try to find the probability for the bigram `(love, natural)`.
    2.  If it's unseen, **"back off"** to a simpler model.
    3.  Just use the probability of the unigram `P(natural)`.
        This is less precise, but it's much better than 0.
  * **How to Calculate and Interpret Perplexity:**
      * **Interpretation:** Perplexity (PPL) is the main way we measure how "good" a language model is. It measures how **surprised** (or confused) the model is by a test sentence. **A lower perplexity is better.**
      * **Analogy:** A perplexity of 20 means that at each word, the model is as confused as if it were trying to guess the next word from a list of 20 equally likely options.
      * **Calculation:** It's the N-th root of the inverse probability of the sentence. $PPL = \left( \frac{1}{P(\text{Sentence})} \right)^{1/N}$


### Why Lower Perplexity Indicates a Better Model
A language model's job is to accurately predict the probability of text.
  * A **low perplexity** means the model assigned a **high probability** to the test sentences. This is good\! It means the model was *not surprised* to see those sentences and that its predictions were accurate. It has learned the language patterns well.
  * A **high perplexity** means the model assigned a **low probability** to the test sentences. This is bad. It means the model was *very surprised* by the real text, and its predictions were wrong.

Therefore, **lower perplexity = less surprised = better predictions = a better model.**

### a) BoW and TF-IDF Vectors
First, we use `scikit-learn` to process the documents, remove common English stop words ("is", "a", "of", "and"), and build the vectors.

````python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# The three documents
documents = [
    "Machine learning is a subset of artificial intelligence", # Doc1
    "Deep learning is a subset of machine learning",          # Doc2
    "Artificial intelligence and machine learning are transforming industries" # Doc3
]
````

````python
# --- 1. Bag-of-Words (BoW) ---
# This just counts the words
bow_vectorizer = CountVectorizer(stop_words='english')
bow_matrix = bow_vectorizer.fit_transform(documents)

# --- 2. TF-IDF ---
# This counts and then weights them by importance
tfidf_vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

# Get the vocabulary (the "bag")
# The columns are: ['artificial', 'deep', 'industries', 'intelligence', 'learning', 'machine', 'subset', 'transforming']
vocab = bow_vectorizer.get_feature_names_out()
````

````python
# --- Displaying the Vectors ---
print("--- Bag-of-Words (BoW) Vectors (Raw Counts) ---")
bow_df = pd.DataFrame(bow_matrix.toarray(), columns=vocab, index=['Doc1', 'Doc2', 'Doc3'])
bow_df
````

````python
print("\n--- TF-IDF Vectors (Weighted Scores) ---")
tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=vocab, index=['Doc1', 'Doc2', 'Doc3'])
# Rounding for simple display
tfidf_df.round(2)
````

*Notice in the BoW vector for Doc2, `learning` has a count of **2**.*

*Notice in the TF-IDF vectors, `learning` and `machine` have the *lowest* scores because they appear in all 3 documents, making them less important.*

### b) Cosine Similarity

We use the matrices from step (a) to calculate similarity.

````python
# --- 1. BoW Similarity ---
bow_sim = cosine_similarity(bow_matrix)
print("\n--- BoW Cosine Similarity ---")
pd.DataFrame(bow_sim, index=['Doc1', 'Doc2', 'Doc3'], columns=['Doc1', 'Doc2', 'Doc3']).round(2)
````

````python
# --- 2. TF-IDF Similarity ---
tfidf_sim = cosine_similarity(tfidf_matrix)
print("\n--- TF-IDF Cosine Similarity ---")
pd.DataFrame(tfidf_sim, index=['Doc1', 'Doc2', 'Doc3'], columns=['Doc1', 'Doc2', 'Doc3']).round(2)
````

### c) Explanations

#### 1\. Why the scores differ:

The similarity scores differ because the two models **weigh words differently**.

  * **BoW** treats all words as equally important. It gives high similarity to Doc1 and Doc2 (0.67) because they both share the words "learning," "machine," and "subset." It doesn't care that "learning" and "machine" are common.
  * **TF-IDF** is smarter. It **punishes common words**. Since "learning" and "machine" appear in all 3 documents, TF-IDF gives them a very low "importance" score. The similarity is now based on the *rarer* shared words. The score for Doc1-Doc2 (0.52) is now mostly based on the word "subset," which is the only unique word they share.

#### 2\. Which representation is better?

| Task | Preferred Method | Why? |
| :--- | :--- | :--- |
| **Information Retrieval** | **TF-IDF** | **Wins.** You want your search to match *important*, *rare* keywords, not just common words. |
| **Document Clustering** | **TF-IDF** | **Wins.** You want to group docs by their unique, defining topics, not just because they all use common words. |
| **Topic Modeling** | **BoW** | **Wins.** Models like LDA (Latent Dirichlet Allocation) are designed to work on raw *counts* (BoW) to discover topics. |

#### 3\. Limitations of Both Approaches

Both BoW and TF-IDF have the same major limitation: **they lose all context.**

  * **No Word Order:** They see "Machine learning" and "learning Machine" as the same thing.
  * **No Semantics:** They have no idea that "artificial intelligence" is related to "deep learning." To them, "deep" and "artificial" are as different as "industries." They can't understand that "big" and "large" mean the same thing.

     - Frequent words

**Hint**: CBOW is faster and works well with frequent words, while Skip-gram is better for rare words and smaller datasets. Consider the number of training instances generated.

## Answer 8: Word2Vec Architectures Deep Dive

### a) CBOW (Continuous Bag-of-Words)
* **Architecture and Training Objective:**
    CBOW is like a "fill-in-the-blank" model. Its architecture is a simple neural network.
    * **Input:** A "bag" (or average) of surrounding context words.
    * **Output:** The single target word in the middle.
    * **Objective:** To predict the middle word based on its context.
* **How it Works:**
    It takes the vectors for all the context words (e.g., "The," "quick," "jumps," "over"), **averages them together** into one single vector, and then uses that average vector to guess the word "fox."
* **Best Use Cases:**
    * **Faster to train.**
    * Works very well for **frequent words**.
    * Slightly better for larger datasets.


### b) Skip-gram
* **Architecture and Training Objective:**
    Skip-gram is the *opposite* of CBOW. It's like a "what's my neighborhood?" model.
    * **Input:** A single target word.
    * **Output:** The surrounding context words.
    * **Objective:** To predict the context words based on the middle word.
* **How it Works:**
    It takes the single vector for "fox" and tries to use it to predict "The," "quick," "jumps," and "over." It has to work harder, as it's predicting multiple words from one.
* **Best Use Cases:**
    * **Better for rare words.** It sees a rare word and is forced to learn a good vector for it to predict its neighbors.
    * **Works very well with small datasets,** as it generates more training examples from a single sentence.


### c) Example and Comparison
**Sentence:** "The quick brown **fox** jumps over the lazy dog"

**Target Word:** "fox"

**Window Size = 2:** (This means we look 2 words to the left and 2 words to the right)

**Context Words:** `[quick, brown, jumps, over]`


#### Training Examples:

* **CBOW (Many-to-One):**
    It creates only **one** training example from this window.
    * **Input:** `(quick, brown, jumps, over)`
    * **Output:** `fox`

* **Skip-gram (One-to-Many):**
    It creates **four** separate training examples from this single window.
    * **Input:** `fox` -> **Output:** `quick`
    * **Input:** `fox` -> **Output:** `brown`
    * **Input:** `fox` -> **Output:** `jumps`
    * **Input:** `fox` -> **Output:** `over`

#### Which Architecture Works Better?

| Condition | Winner | Why? |
| :--- | :--- | :--- |
| **Small Datasets** | **Skip-gram** | It creates far more training examples from the same amount of text (as seen in the example above). |
| **Rare Words** | **Skip-gram** | It is forced to learn a good vector for the rare word to predict its context. <br>CBOW just averages the context and doesn't learn as much about the rare word itself. |
| **Frequent Words** | **CBOW** | It's faster. It quickly learns good vectors for common words by averaging their many different contexts, which is more efficient. |

Compare and contrast GloVe and FastText embedding techniques:

   - How does GloVe use global co-occurrence statistics?
   - How does FastText incorporate subword information?

b) **Handling Out-of-Vocabulary (OOV) words**:
   - Given the trained words: "playing", "player", "played"
   - How would each model handle the unseen word "gameplay"?
   - Which model is more suitable for morphologically rich languages (e.g., German, Turkish)?

c) **Practical considerations**:
   - Training time and computational requirements
   - Model size and memory footprint
   - Performance on rare and misspelled words

**Hint**: FastText breaks words into character n-grams (e.g., "playing" → "<pl", "pla", "lay", "ayi", "yin", "ing", "ng>"). GloVe uses matrix factorization on co-occurrence counts.

## Answer 9: GloVe vs FastText Comparison

### a) Training Methodology
* **How GloVe works (Global Co-occurrence):**
    1.  First, it scans the *entire* corpus to build a giant **co-occurrence matrix**.
    2.  This matrix stores how many times each word (e.g., "ice") appears in the context of every other word (e.g., "steam").
    3.  GloVe's objective is to learn vectors for each word so that the **dot product** of any two word vectors equals the (log of) their co-occurrence count. It's a "count-based" model.
* **How FastText works (Subword Information):**
    1.  It's an extension of Word2Vec (Skip-gram). It doesn't treat words as single units.
    2.  Instead, it breaks every word into **character n-grams** (e.g., "playing" becomes `<pl`, `pla`, `lay`, `ayi`, `yin`, `ing`, `ng>`).
    3.  It learns a vector for *each n-gram*.
    4.  The final vector for a word (like "playing") is simply the **sum** of the vectors of all its n-grams. It's a "predictive" model.



### b) Handling Out-of-Vocabulary (OOV) Words

Let's use the example: trained on "playing," "player," and "played," but not on the new word "gameplay."
* **How GloVe handles "gameplay":**
    * It fails. "gameplay" was not in its vocabulary during training. GloVe has no vector for it.
    * It will assign a special **`<UNK>` (unknown) token** or a random vector. This vector has no meaning and is useless.
* **How FastText handles "gameplay":**
    * It succeeds. FastText breaks the new word "gameplay" into its n-grams (e.g., `<ga`, `gam`, `ame`, `mep`, `epl`, `pla`, `lay`, `ay>`).
    * It *has* seen the n-grams `pla`, `lay`, and `ay` before (from "playing," "player," etc.).
    * It **sums the vectors of these known n-grams** to create a brand-new, high-quality vector for "gameplay" on the fly. This new vector will be meaningfully close to the vectors for "playing" and "player."
* **Which is better for morphologically rich languages (German, Turkish)?**
    * **FastText, by far.** Languages like Turkish can have hundreds of forms for a single root word. GloVe would see these as all-new, unknown words. FastText recognizes they all share the same root n-grams and can create good vectors for all of them.



### c) Practical Considerations
| Feature | GloVe | FastText |
| :--- | :--- | :--- |
| **Training Time** | **Faster.** It just counts and then trains on the matrix. | **Slower.** It has to train on *billions* of character n-grams. |
| **Model Size** | **Smaller.** Stores one vector per *word* in its vocabulary. | **Much Larger.** Must store a vector for *every n-gram*. |
| **Rare Words** | **Poor.** Rare words have few stats, so their vectors are bad. | **Excellent.** Can build a good vector from a word's parts, even if it's rare. |
| **Misspelled Words** | **Fails.** A typo (e.g., "playinng") is just another OOV word. | **Excellent.** A typo still shares most n-grams with the correct word, so its vector is very similar. |

## Answer 10: Classical vs Distributed Representations - Application Perspective

### 1. Legal Document Search Engine
* **a) Choice:** **Classical (TF-IDF)**
* **b) Justification:**
    * **Semantic Similarity:** Low requirement. Legal search is about **exact keyword matching**. "Breach of contract" and "contract violation" are not semantically identical in a legal context.
    * **Vocabulary:** Highly specific, domain-specific jargon (e.g., "indemnify," "heretofore"). TF-IDF excels at finding these exact, rare terms.
    * **Interpretability:** **High requirement.** A lawyer *must* know *why* a document was retrieved. TF-IDF is perfectly interpretable—it can show exactly which keywords matched.
    * **Training Data:** Legal data is often private, making it hard to train good, specific embeddings. TF-IDF needs no special training data.


### 2. Chatbot Intent Classification
* **a) Choice:** **Distributed (Word2Vec / FastText)**
* **b) Justification:**
    * **Semantic Similarity:** **High requirement.** The *meaning* of the query is everything. "Book a flight," "I need a ticket," and "fly to Boston" must all map to the same `BookFlight` intent.
    * **Vocabulary:** Informal, broad, and full of variations, slang, and typos. **FastText** is especially good here because its subword information can handle typos ("flght") and new words.
    * **Interpretability:** Low. The developer just cares that the correct intent was classified, not *why* the vector math worked.
    * **Training Data:** Chat logs provide a rich source of data to train or fine-tune embeddings.


### 3. Academic Paper Recommendation System
* **a) Choice:** **Distributed (Word2Vec / GloVe)**
* **b) Justification:**
    * **Semantic Similarity:** **High requirement.** The goal is to find *related* papers. A paper on "Neural Networks" should be recommended to someone reading about "Deep Learning." TF-IDF would miss this connection, but distributed embeddings are designed to capture it.
    * **Vocabulary:** Specific, but the *concepts* are what matter.
    * **Training Data:** Large public datasets (like arXiv, PubMed) are available to pre-train very strong, domain-specific embeddings (e.g., SciBERT, or just Word2Vec trained on a science corpus).
    * **Interpretability:** Medium. It's helpful, but the quality of the recommendation (semantic relevance) is more important.


### Hybrid Approaches

Yes, hybrid approaches can significantly improve performance by combining the strengths of both.

**How?**
1.  **As Combined Features:** You can create both a TF-IDF vector and an averaged Word2Vec vector for a document. Then, you concatenate these two vectors and feed them into a machine learning model (like a classifier or a ranking model). This gives the model both **keyword precision (from TF-IDF)** and **semantic understanding (from Word2Vec)**.

2.  **As a Two-Stage System:** This is very common in search.
    * **Stage 1 (Retrieve):** Use the very fast **TF-IDF** to find the top 100 documents that match the user's keywords.
    * **Stage 2 (Re-rank):** Use a slower, more "expensive" **distributed model** to calculate the semantic similarity for just those 100 documents and re-rank them for the user.

For the **academic paper system**, this hybrid approach is ideal. It finds papers with the right *keywords* (TF-IDF) and then ranks them by *conceptual similarity* (Word2Vec/GloVe).
